# Tabulková data CSV

Modul `csv` — práce s CSV soubory (comma separated values)
- `DictReader`: načte tabulku (v CSV souboru) do slovníku
- `DictWriter`: zapisuje slovník do tabulky (v CSV souboru)

```python
import csv

with open("./soubory/skore.csv", encoding="utf-8", newline="") as soubor:
    tabulka = csv.DictReader(soubor)            # Připraví se kurzor

    for radek in tabulka:
        jmeno = radek["jmeno"]                  # Až teď se kurzor posouvá
        skore = int(radek["skore"])
        print(f"{jmeno}: {skore} bodů")
```

### Zápis

Zvlášť se zapisují hlavičky a zvlášť jednotlivé řádky.

```python
import csv

hraci = [
    {"jmeno": "Ada", "skore": 120},
    {"jmeno": "Bořek", "skore": 95},
    {"jmeno": "Cyril", "skore": 180}
]
sloupce = ["jmeno", "skore"]

with open("./soubory/skore-novy.csv", mode="w", encoding="utf-8", newline="") as soubor:
    tabulka = csv.DictWriter(soubor, fieldnames=sloupce, delimiter=";")
    tabulka.writeheader()
    tabulka.writerows(hraci)
```

**Mikroůloha** Najdi hráče s nejvyšším skóre a převeď hodnoty skóre na celá čísla (použij skore-novy.csv).

```python
import csv

hodnoty = {}
with open("./soubory/skore-novy.csv", mode="r", encoding="utf-8", newline="") as soubor:
    tabulka = csv.DictReader(soubor, delimiter=";")
    hodnoty = {radek["jmeno"]: int(radek["skore"]) for radek in tabulka}

nejlepsiHrac = ""
for jmeno, skore in hodnoty.items():
    if skore > hodnoty.get(nejlepsiHrac, 0):
        nejlepsiHrac = jmeno

print(f"{nejlepsiHrac} má největší skóre {hodnoty.get(nejlepsiHrac, 0)}")
```
