## Binární soubory, obrázky a JSON

Binární režim — data se čtou jako bajty.

JSON — strukturovaný textový soubor, hodí se na ukládání např. slovníku, seznamů apod. (typicky konfigurace).

```python
with open("./soubory/obrazek.png", mode="rb") as soubor:
    obrazek_data = soubor.read()

print(f"Velikost obrázku: {len(obrazek_data)} bajtů = {len(obrazek_data) / 1024} kibibajtů")
print(f"Prvních 8 bajtů:", obrazek_data[:8])
```

```python
with open("./soubory/kopie_obrazku.png", mode="wb") as soubor:
    soubor.write(obrazek_data)
```

Začátek binárního souboru je 8-bajtová signatura, která je ro daný typ vždy stejná.

```python
png_signatura = b"\x89PNG\r\n\x1a\n"
print("Je to PNG", obrazek_data.startswith(png_signatura))
```

### Ukázka zobrazení obrázku

```python
import tkinter as tk

okno = tk.Tk()
okno.title("Obrázek na plátně")

platno = tk.Canvas(okno, width=400, height=300, background="white")
platno.pack()

obrazek = tk.PhotoImage(file="./soubory/obrazek.png")
platno.create_image(20, 20, anchor="nw", image=obrazek)

okno.mainloop()
```

### JSON

Java Script Object Notation

```python
stav_hry = {
    "zivoty": 3,
    "batoh": ["meč", "pochodeň"]
}
```

```python
import json

# Zápis do JSON souboru
with open("./soubory/stav_hry.json", mode="w", encoding="utf-8") as soubor:
    json.dump(stav_hry, soubor, ensure_ascii=False, indent=2)

# Čtení z JSON souboru
with open("./soubory/stav_hry.json", mode="r", encoding="utf-8") as soubor:
    nacteny_stav = json.load(soubor)

print(nacteny_stav)
```
